```python
#Convert date of registration on datetime format
df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'])
```

```python
df['Income']=(pd.to_numeric(df['Income']))
```

```python
#Which Marital_status has the most monthly income
t = df.groupby('Marital_Status').agg({'Income': ['min', 'mean', 'max']})
  
print("Min, mean, and max Income of each type of marital Status : \n")
print(t)
#plt.figure(figsize=(9,9)
t.plot(kind='bar')
```

```python
df['Age'] = df.apply(lambda df:  2021- df['Year_Birth'], axis=1)
df['Spent'] = df.apply(lambda df: df['MntFishProducts'] + df['MntFruits'] + df['MntGoldProds'] + 
                df['MntMeatProducts'] + df['MntSweetProducts'] + df['MntWines'], axis=1)


df['Children'] = df.apply(lambda df: df['Kidhome'] + df['Teenhome'], axis=1)

df['TotalAcceptedCmp'] = df.apply(lambda df: df['AcceptedCmp1'] + df['AcceptedCmp2'] + 
                                      df['AcceptedCmp3'] + df['AcceptedCmp4'] + 
                                      df['AcceptedCmp5'], axis=1)
df['NumTotalPurchases'] = df.apply(lambda df: df['NumWebPurchases'] + df['NumCatalogPurchases'] + 
                            df['NumStorePurchases'] + df['NumDealsPurchases'], axis=1)
```

```python
# Drop'Year_Birth', 'Z_CostContact', 'Z_Revenue', 'Kidhome', 'Teenhome' columns
df = df.drop(['Year_Birth', 'Z_CostContact', 'Z_Revenue', 'Kidhome', 'Teenhome'], axis=1) 
```

```python
df['Age']
```

```python
df = df.drop(df.loc[df['Income'] > 600000].index)
df = df.drop(df.loc[df['Age'] > 100].index)
```

```python
# Cut the continuous variable Age
bins = [0, 20, 40, 60, 80,100]
aged = pd.cut(df['Age'], bins=bins)
print('number of age_cut:\n', aged.value_counts())
```

```python
# Different age group customer classification summary
plt.figure(figsize=(12,6))
sb.countplot(y=aged, palette='Set2')
plt.title('countplot of Age')
```

```python
# The relationship between education level and shopping expenses
# Comparison of customer spending in different marital status

to_boxplot = ['Education', 'Marital_Status']
fig, axes = plt.subplots(2, 1, figsize=(12, 12))
axes = axes.flatten()

for col, ax in zip(to_boxplot, axes):
    ax = sb.boxplot(data=df, x=col, y='Spent', ax=ax, palette='Set2')
    ax.set_title(f'boxplot of spent by {col}')
```

```python
# Statistics of monthly activity frequency of website users
plt.figure(figsize=(12,6))
plt.hist(df['NumWebVisitsMonth'], bins=50)
plt.title('Monthly activities (5-7)/month')
```

```python
# Website complaints
x = df['Complain'].value_counts().sort_values()

plt.figure(figsize=(8,8))
plt.pie(x=x, labels=['YES', 'NO'], autopct = '%1.1f%%')
plt.title('Complain of customer')
```

```python
# Frequency statistics of 'MntWines', 'MntFishProducts', 'MntFruits', 'MntGoldProds', 'MntMeatProducts', 'MntSweetProducts'  respective sales
to_histplot = ['MntWines', 'MntFishProducts', 'MntFruits', 'MntGoldProds', 
               'MntMeatProducts', 'MntSweetProducts']

fig, axes = plt.subplots(3, 2, sharex=True, sharey=True, figsize=(14, 10))
axes = axes.flatten()

for col, ax in zip(to_histplot, axes):
    ax = plt.hist(data=df, x=col, ax=ax)
    ax.set_title(f'histogram of {col}')
```

```python
# Website monthly active users line chart
groupedDate = df.set_index('Dt_Customer')
groupedDate = groupedDate.resample('M').count()

plt.figure(figsize=(14,4))
plt.title('Customer NumWebVisitsMonth by mounths')
groupedDate.NumWebVisitsMonth.plot(kind='line')
```

```python
# Line chart of users purchased through the catalog each month
groupedDate = df.set_index('Dt_Customer')
groupedDate = groupedDate.resample('M').count()

plt.figure(figsize=(14,4))
plt.title('Customer NumCatalogPurchases by mounths')
groupedDate.NumCatalogPurchases.plot(kind='line')
```
